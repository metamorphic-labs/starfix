```python
import pyarrow as pa
from starfix import ArrowDigester

# Create a simple Arrow table
data = {
    "id": [1, 2, 3, 4, 5],
    "name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "value": [10.5, 20.3, 15.7, 30.2, 25.8],
}
table = pa.table(data)

# Hash the entire Arrow table (every chunk)
print(ArrowDigester.hash_table(table).hex())

# Hash the schema of the Arrow table
print(ArrowDigester.hash_schema(table.schema).hex())

# Many tables of one schema: analyse the schema once
prepared = ArrowDigester.prepare(table.schema)
print(prepared.hash_table(table).hex())
```
